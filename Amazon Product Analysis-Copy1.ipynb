```python
#check missing values in all attributes
missing_values_all = asales.isnull().sum()
print("Missing values for each attribute:")
print(missing_values_all)
```

```python
#Check for missing values in the 'rating_count' column and fill them with the median value.
missing_values = asales['rating_count'].isnull().sum()
print("Number of missing values in 'rating_count' column:", missing_values)
```

```python
#Remove commas from 'rating_count' column and convert to numeric
asales['rating_count'] = asales['rating_count']
.str.replace(',', '').astype(int)

#Fill missing values with the median value
median_rating_count = asales['rating_count'].median()
asales['rating_count'].fillna(median_rating_count, inplace=True)
```

```python
#Verify that missing values have been filled
missing_values_after_fill = asales['rating_count'].isnull().sum()
print("Number of missing values in 'rating_count' column after filling:", missing_values_after_fill)
```
