# E. Data Visualization

1. Distribution of Purchase Price

```python
plt.figure (figsize=(10, 10))
sns.histplot(ecom['Purchase Price'], kde=True)
plt.title('Distribution of Purchase Price')
plt.xlabel('Purchase Price')
plt.ylabel('Frequency')

#show the plot
plt.show()
```

2. Customer Language Distribution

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Language', data=ecom)
plt.title('Customer Language Distribution')
plt.xlabel('Language')
plt.ylabel('Count')
plt.show()
```

 3. Preferred Credit Card Providers

```python
plt.figure(figsize=(12, 6))
sns.countplot(x='CC Provider', data=ecom,
              order=ecom['CC Provider'].value_counts().index)
plt.title('Preferred Credit Card Providers')
plt.xlabel('Credit Card Provider')
plt.ylabel('Count')
plt.show()
```

4. Preferred Hour of Purchase

```python
unique_exp_dates = ecom['CC Exp Date'].unique()
print("Unique values in 'CC Exp Date' column:", unique_exp_dates)

# Try to convert 'CC Exp Date' with explicit format
ecom['CC Exp Date'] = pd.to_datetime(ecom['CC Exp Date'], errors='coerce',format='%m/%y')

# Check for missing values after conversion
print("Missing values after conversion:", ecom['CC Exp Date'].isnull().sum())

ecom['Hour of Purchase'] = ecom['CC Exp Date'].dt.hour
```

```python
# Visualization of 'Hour of Purchase'
plt.figure(figsize=(3, 6))
sns.countplot(x='Hour of Purchase', data=ecom)
plt.title('Preferred Hour of Purchase')
plt.xlabel('Hour of Purchase')
plt.ylabel('Count')
plt.show()
```

 5. Preferred Website for Advertisements (Top 5)

```python
plt.figure(figsize=(14, 6))
sns.countplot(x='Browser Info', data=ecom, order=ecom['Browser Info'].value_counts().index[:5])
plt.title('Preferred Website for Advertisements (Top 5)')
plt.xlabel('Browser Info')
plt.ylabel('Count')
plt.show()
```
